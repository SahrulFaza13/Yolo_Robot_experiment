# 02 - Training model YOLO

Transfer learning dari model pretrained. Hasil terbaik ada di
`runs/detect/person_model/weights/best.pt`.

```python
%load_ext autoreload
%autoreload 2

import torch
from ultralytics import YOLO

print("CUDA tersedia:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```

```python
model = YOLO("yolo11n.pt")      # pretrained COCO -> transfer learning

model.train(
    data="dataset/person/data.yaml",
    epochs=100,
    imgsz=640,
    batch=8,                     # kecilkan jadi 4 kalau CUDA out of memory
    patience=20,
    name="person_model",
    device=0,                    # 0 = GPU, "cpu" = CPU
)
```

## Evaluasi model terbaik

```python
best = YOLO("runs/detect/person_model/weights/best.pt")
metrics = best.val(data="dataset/person/data.yaml")
print("mAP50:", metrics.box.map50)
print("mAP50-95:", metrics.box.map)
```

```python
from robot.command_link import CommandLink

PORT = "COM5"          # sesuaikan dengan hasil di atas
link = CommandLink(PORT, baud=115200)
print("Terhubung ke", PORT)
```

```python
import time

for nama, cmd in [("maju", "1"), ("belok kiri", "2"),
                  ("belok kanan", "3"), ("mundur", "4"), ("stop", "0")]:
    print(nama, "->", cmd)
    link.send(cmd)
    time.sleep(1.5)
link.send("0")
```
